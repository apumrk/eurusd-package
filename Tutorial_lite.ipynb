### 1. Data Loading & Preprocessing 
#### Initialize the data loader and load your dataset:

To ensure the `DataLoader` can successfully load your data, please follow these steps:

1. **Create a Directory**: Navigate to the working directory of your project where your scripts or Jupyter notebooks are located. Inside this directory, create a new folder named `data`.

2. **Add Data File**: Place your data file, which by default should be named `EURUSD_data.csv`, into the `data` folder. If you use a different filename or have multiple data sets, make sure to specify the correct file name when initializing the `DataLoader`.

3. **Initialize DataLoader**: When you set up `DataLoader` in your script or notebook, it will automatically look for the data file in the `data` subdirectory. Here’s how you can initialize the `DataLoader`:

Method: preprocess_data

The preprocess_data method ensures that the date column is converted to the correct format and that the data is cleaned and ready for analysis.

````python
loader = DataLoader()
loader.preprocess_data()
data = loader.get_data()
````

### 2. Data Analyzing with Basic Information
#### Create an instance of DataAnalyzer to examine basic statistical information about the dataset:

Method: describe_data

The describe_data method returns statistical summaries such as mean, standard deviation, and percentiles of the numerical columns in the DataFrame.

````python
analyzer = DataAnalyzer(data)
print(analyzer.describe_data())
````

Method: describe_info

The describe_info method gives a concise summary of the DataFrame, including column data types, non-null counts, and memory usage.

````python
print(analyzer.describe_info())
````

### 3. Data Visualization with it's Price, Trend, Residual and Seasonality Components Separately
#### Visualize different aspects of the data using DataVisualizer:

Method: plot_time_series

This method plots the original time series data, showing how the values change over time.

````python
visualizer = DataVisualizer(data)
visualizer.plot_time_series()
````

Method: plot_Trend_Component

The plot_Trend_Component method visualizes the trend component of the time series data.

````python
visualizer.plot_Trend_Component()
````

Method: plot_Seasonal_Component

The plot_Seasonal_Component method visualizes the seasonal component of the data.

````python
visualizer.plot_Seasonal_Component()
````

Method: plot_Residual_Component

The plot_Residual_Component method visualizes the residual (or noise) component of the data.

````python
visualizer.plot_Residual_Component()
````
